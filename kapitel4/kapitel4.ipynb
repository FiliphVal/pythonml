## Uppgift 1
Klassificeringsproblem handlar om att förutsäga vilken diskret klass eller kategori en observation tillhör.
Exempel: Spamfilter för e-post, medicinsk diagnos (sjuk/frisk), bildigenkänning av siffror eller kundavhopp.

## Uppgift 2
- OvR (One-vs-Rest): Tränar en binär modell för varje klass mot alla andra klasser tillsammans. Vid prediktion väljs den klass som får högst sannolikhet.
- OvO (One-vs-One): Tränar en binär modell för varje unikt par av klasser. Modellerna röstar och klassen med flest röster vinner.

## Uppgift 3
Förklaring av utvärderingsmått:
a) Confusion matrix: En tabell som visar rätt och fel klassificeringar uppdelat i True Positives, False Positives, True Negatives och False Negatives.
b) Accuracy: Andelen korrekta prediktioner av det totala antalet observationer.
c) Precision: Andelen av alla positiva prediktioner som verkligen var korrekta.
d) Recall: Andelen av alla verkligt positiva fall som modellen lyckades hitta.
e) F1-score: Det harmoniska medelvärdet mellan precision och recall.
f) ROC-kurvan: Visar sambandet mellan sanna positiva och falska positiva vid olika beslutströsklar.

## Uppgift 4
Precision-recall trade-off innebär att om vi ändrar beslutströskeln för att få högre precision (färre falsklarm) så missar modellen ofta fler verkliga fall och får lägre recall, och vice versa.

## Uppgift 5
Översikt av klassificeringsmodeller:
a) Logistisk regression: Beräknar sannolikheten för en klass med en sigmoidfunktion.
b) Support vector machines (SVC): Hittar den linje eller det plan som ger störst marginal mellan klasserna.
c) Beslutsträd: Delar upp datan steg för steg med if/else-villkor.
d) Ensemble learning: Kombinerar flera modeller för säkrare förutsägelser.
e) Random forest: En samling av många slumpade beslutsträd som röstar om resultatet.
f) Extra trees: Trädmodell som slumpar även tröskelvärdena vid delningarna för att minska överanpassning.

## Uppgift 6
Feature importance i trädmodeller visar hur mycket varje enskild variabel har bidragit till att göra delningar i träden. Det hjälper oss att förstå vilka variabler som är viktigast för prediktionen.

## Uppgift 7
När man strävar efter högsta möjliga precision minskar i regel recall.
- Hög precision önskas när falska positiva är kostsamma eller farliga (t.ex. skräppostfilter där viktiga mail inte får raderas).
- Hög recall önskas när det är kritiskt att inte missa något positivt fall (t.ex. vid allvarliga sjukdomar).
- I rättsväsendet prioriteras mycket hög precision för att undvika att oskyldiga döms till fängelse.

## Uppgift 8
Figur 4.8 visar hur precision och recall varierar när beslutströskeln ändras. Höjs tröskeln blir modellen mer försiktig, vilket ger högre precision men lägre recall. Sänks tröskeln fångas fler fall (högre recall) men fler falsklarm uppstår (lägre precision).

## Uppgift 11
Förklara koden och tolka classification_report.

```python
from sklearn.metrics import classification_report

y_true = [0, 1, 2, 2, 2]
y_pred = [0, 0, 2, 2, 1]
target_names = ['class 0', 'class 1', 'class 2']
print(classification_report(y_true, y_pred, target_names=target_names))
```

Koden beräknar precision, recall och F1-score per klass samt övergripande accuracy:
- Class 0: Gissades 2 gånger varav 1 var rätt (precision 0.50). Den enda sanna nollan hittades (recall 1.00).
- Class 1: Modellen missade den sanna ettan (precision 0.00, recall 0.00).
- Class 2: Gissades 2 gånger och båda var rätt (precision 1.00). Två av tre sanna tvåor hittades (recall 0.67).
- Totalt: Accuracy är 0.60 (3 rätt av 5 observationer).

## Uppgift 13
Klassificeringsmodell på hr_employee_data.xlsx för att förutsäga left.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

df = pd.read_excel('../data/hr_employee_data.xlsx')

y = df['left']
X = df.drop(columns=['Emp_Id', 'left'])
X = pd.get_dummies(X, drop_first=True)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

model = LogisticRegression(max_iter=1000)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
print(f"Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print("\nKlassificeringsrapport:")
print(classification_report(y_test, y_pred))
```

## Uppgift 15
Klassificering av handskriven siffra från egen bild.

```python
import numpy as np
from PIL import Image, ImageOps
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# 1. Träna modell på digits-datan
digits = load_digits()
X = digits.data
y = digits.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
model = LogisticRegression(max_iter=1000, random_state=42)
model.fit(X_train, y_train)
print(f"Modellens test-accuracy: {accuracy_score(y_test, model.predict(X_test)):.4f}")

# 2. Läs in och preprocessa egen bild
img = Image.open('min_siffra.png').convert('L')
img = img.resize((8, 8), Image.Resampling.LANCZOS)
img_inv = ImageOps.invert(img)
img_array = np.array(img_inv) / 255.0 * 16.0

pred = model.predict(img_array.reshape(1, -1))
print(f"Predikterad siffra: {pred[0]}")
```
