1. Hur hänger AI, ML och DL ihop?
Kan ses som ryska dockor där varje del ryms i den andra, all djupinlärning är AI, all maskinlärning är AI men AI är varken maskininlärning eller djupinlärning.
2. Vilka är de fyra problemkategorierna inom ML?
Klassificering, Regression, Klustring och dimensionsreducering
3. a) Vad är syftet med att dela upp data i träningsdata, valideringsdata och testdata?
För att undvika att sin modell tränar på samma data, för då är risken att den memorerat datan.
b) Vad är k-delad korsvalidering för något?
Så man kan dela datan i olika blocks, så risken inte finns att vi hade tur eller otur med datan.
c) Vad är root mean squared error (RMSE) för något?
Ett mått som mäter det genomsnittliga felet för o veta hur mycket fel modellen görs. Men innan snittet dras så kvaderas alla fel, för att straffa extremt stora avikelser och för att alla negativa tecken försvinner.
d) Vad är en hyperparameter för något? Vad är en parameter för något?
Parameter : Interna vikter och variabler som modellen räknar ut själv utifrån träningsdatan.
Hyperparameter: Modellens inställningar som vi människor måste sätta innan träningen börjar, exempelvis hur många grenar får ett beslutsträd ha.
e)Vad är grid search och hur hänger namnet “grid” och “search” ihop med processen som genomförs? Vad innebär hyperparametern refit=True?
Vi ger programmet ett rutnätt av alternativ, exempelvis träddjup på 3, 5 och 10, sedan kommer gridsearch söka igenom alla möjliga kombinationer och bygga en modell för varje.
refit=True kommer när den bästa kombinationen är funnen att ta denna kombination och träna en helt ny slutglitig modell på träningsdatasetet.
f)Vad är kategorisk data och hur hanteras det? Använd begreppen nominal data, ordinal data, one-hot-encoding, dummy-variableencoding och ordinal encoding.
Kategorisk är variabler som representerar grupper och skiljs i två typer. Nominal och ordinal data.
ORdinal data är kategorier som har en naturlig och logisk ordning, exempelvis klädstorlekar S, M, L. Ordinal Encoding byter ut värdena mot siffror S=1 M=2 3=L och algoritmen förstår att 3 är större än 1.
Nominal Data är kategorier som saknar inbördes ordning, exempelvis bilmärken. En volvo är inte mer eller mindre än en honda. Om vi ger volvo 1 och bmw 3 skulle algoritmen tro att en bmw är 3 gånger så mycket mer värt än en volvo. Därför använder vi one hot encoding, som skapar kolumner så som Är_Volvo, Är_bmw, om det är en volvo blir värdet 1 och annars 0.
g) Vad är feature engineering för något?
Använda sin domänkunskap för att skapa nya kolumner ur befintlig data. T.ex data över försäljning med datum 2026-12-18 så kan de va svårt o se mönster. Genom feature enginering kan du extrahera veckodag, måndag eller exempelvis Är_Helgdag.
h) Vad menas med principle of parsimony?
Om vi har två modeller, en jätteavancerad och en simpe, men båda presterar ungefär lika bra ska vi alltid välja den lättare modellen. Pga den är snabbare, lättare att tolka och rymmer färre okända fel.
4. Vad menas med att “en modell är en förenkling av verkligheten”?
Den visar bara det som är viktigt för att förstå en viss sak. En karta är till exempel en modell som visar vägar, men den visar inte varje träd eller sten
5. Vad menas med att en modell är “överanpassad” eller overfitted på engelska?
När vi låter en modell bli för komplex och tränat för länge, den slutar då leta efter de generella mönstren och börjar memorera träningsdatans exakta utseende. Kommer ha 100% rätt på träningsdatan men när den får se ny data i verkligheten kommer den tappa det då den tappat förmågan att generalisera. Det är här principle of parsimony kommer in i bilden.
6. Högre är bättre i scikit-learn scoring, vad innebär det?
För att det är bättre resultat. sklearn använder neg_mean_squared_error och tar felet och sätter ett minustecken framför det, ett fel på 100 blir därför -100, därför är -5 bättre än -100.
7. Vad är tvärsnittsdata, tidsseriedata och paneldata? Exemplifiera när respektive datakategori kan uppstå
Tvärsnittsdata - Ögonblicksbild, möter flera olika subjet vid exakt samma tidpunkt. Exempelvis data på alla svenskars inkomster år 2025.
Tidsseriedata - Följer ett subjekt över tid, exempelvis priset på en specifik aktie mätt varje dag under fem års tid.
Paneldata - Kombination av dem båda, exempelvis BNP utvecklingen i 20 olika lönder över 30 års tid.
