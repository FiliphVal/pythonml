## Uppgift 3
a) Syfte med uppdelning i tränings-, validerings- och testdata: För att modellen inte ska memorera träningsdatan. Träningsdata används för att träna modellen, valideringsdata för att välja modell och justera inställningar, och testdata sparas till sist för att utvärdera hur modellen klarar helt ny data.

b) k-delad korsvalidering: Datan delas in i k stycken delar (block). Modellen tränas på k-1 delar och testas på den sista, vilket upprepas k gånger så alla delar får vara testdata en gång. Detta minskar risken för att ett slumpmässigt urval blir missvisande.

c) RMSE (Root Mean Squared Error): Ett mått på modellens genomsnittliga fel i samma enhet som det vi förutsäger. Felen kvadreras först för att straffa stora avvikelser extra hårt och ta bort minustecken, och sedan dras roten ur medelvärdet.

d) Hyperparameter vs parameter:
- Parameter: Modellens interna vikter som den räknar ut själv från träningsdatan.
- Hyperparameter: Inställningar som vi väljer innan träningen startar, t.ex. maximalt djup i ett beslutsträd.

e) Grid search och refit=True:
- Grid search testar systematiskt alla kombinationer av hyperparametrar i ett rutnät för att hitta den bästa uppsättningen.
- refit=True innebär att modellen automatiskt tränas om på hela träningsdatasetet med de bästa parametrarna när sökningen är klar.

f) Kategorisk data:
- Nominal data: Kategorier utan inbördes ordning (t.ex. bilmärken som Volvo, BMW). Hanteras med one-hot encoding eller dummy variable encoding (drop_first=True) som skapar 0/1-kolumner.
- Ordinal data: Kategorier med en naturlig ordning (t.ex. storlekar S, M, L). Hanteras med ordinal encoding där värdena ersätts med siffror (1, 2, 3).

g) Feature engineering: Att använda domänkunskap för att skapa nya variabler ur befintlig data, t.ex. att plocka ut veckodag eller om det är helgdag ur ett datum.

h) Principle of parsimony (Ockhams rakkniv): Om två modeller presterar ungefär lika bra bör man välja den enklare modellen eftersom den är snabbare, lättare att förstå och har mindre risk för oväntade fel.

## Uppgift 4
Vad menas med att "en modell är en förenkling av verkligheten"?
Modellen visar bara de viktigaste sambanden för att förstå eller förutsäga något. En karta är till exempel en förenkling som visar vägar, men den visar inte varje enskilt träd eller sten.

## Uppgift 5
Vad menas med att en modell är överanpassad (overfitted)?
Att modellen är för komplex och har lärt sig träningsdatans brus och detaljer utantill istället för det generella mönstret. Den får väldigt bra resultat på träningsdata men presterar dåligt på ny data.

## Uppgift 6
Högre är bättre i scikit-learn scoring, vad innebär det?
Scikit-learn är byggt så att ett högre poäng alltid tolkas som bättre resultat. Därför görs felmått som MSE om till negativa tal (t.ex. neg_mean_squared_error), så att ett litet fel som -2 räknas som bättre än ett stort fel som -10.

## Uppgift 7
Tvärsnittsdata, tidsseriedata och paneldata:
- Tvärsnittsdata: Mätningar på flera olika subjekt vid en och samma tidpunkt (t.ex. inkomst för 1000 personer år 2025).
- Tidsseriedata: Mätningar på samma subjekt över tid (t.ex. priset på en aktie varje dag under fem år).
- Paneldata: Kombination av båda där flera subjekt följs över tid (t.ex. BNP för 20 länder under 30 års tid).
